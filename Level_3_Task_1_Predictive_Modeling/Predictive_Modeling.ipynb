```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV
```

# Objective

The objective of this task is to build and evaluate machine learning classification models for predicting customer churn.

The project includes:
- Loading the training and testing datasets.
- Preprocessing the data.
- Encoding categorical variables.
- Scaling numerical features.
- Training multiple classification models.
- Evaluating model performance.
- Performing hyperparameter tuning using GridSearchCV.

```python
train_df = pd.read_csv("churn_train.csv")

train_df.head()
```

```python
test_df = pd.read_csv("churn_test.csv")

test_df.head()
```

# Dataset Information

Dataset Used:
- churn_train.csv
- churn_test.csv

Target Variable:
- Churn

This dataset contains customer information such as account details, service usage, call charges, and customer churn status.

```python
print("Training Dataset Shape:", train_df.shape)
print("Testing Dataset Shape:", test_df.shape)

train_df.info()
```

# Data Preprocessing

The following preprocessing steps were performed:

- Loaded training and testing datasets.
- Checked dataset structure.
- Encoded categorical columns using LabelEncoder.
- Separated features and target variable.
- Standardized numerical features using StandardScaler.

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

# Encode categorical columns
for col in ["State", "International plan", "Voice mail plan", "Churn"]:
    train_df[col] = le.fit_transform(train_df[col])
    test_df[col] = le.transform(test_df[col])

train_df.head()
```

```python
# Features and Target
X_train = train_df.drop("Churn", axis=1)
y_train = train_df["Churn"]

X_test = test_df.drop("Churn", axis=1)
y_test = test_df["Churn"]

print(X_train.shape)
print(X_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

# Model Training

Three machine learning classification models were trained:

- Logistic Regression
- Decision Tree Classifier
- Random Forest Classifier

These models were trained using the preprocessed training dataset.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

models = {
    "Logistic Regression": LogisticRegression(max_iter=1000),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "Random Forest": RandomForestClassifier(random_state=42)
}

trained_models = {}

for name, model in models.items():
    model.fit(X_train_scaled, y_train)
    trained_models[name] = model
    print(f"{name} trained successfully!")
```

```python
# Make predictions

lr_pred = trained_models["Logistic Regression"].predict(X_test_scaled)
dt_pred = trained_models["Decision Tree"].predict(X_test_scaled)
rf_pred = trained_models["Random Forest"].predict(X_test_scaled)
```

# Model Evaluation

The trained models were evaluated using the following performance metrics:

- Accuracy
- Precision
- Recall
- F1-Score

These metrics were used to compare the performance of each classification model.

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

predictions = {
    "Logistic Regression": lr_pred,
    "Decision Tree": dt_pred,
    "Random Forest": rf_pred
}

for name, pred in predictions.items():
    print("="*40)
    print(name)
    print("Accuracy :", accuracy_score(y_test, pred))
    print("Precision:", precision_score(y_test, pred))
    print("Recall   :", recall_score(y_test, pred))
    print("F1 Score :", f1_score(y_test, pred))
```

# Hyperparameter Tuning

GridSearchCV was used to optimize the Random Forest model.

Parameters Tuned:
- n_estimators
- max_depth

The best parameter combination was selected based on cross-validation accuracy.

```python
from sklearn.model_selection import GridSearchCV

param_grid = {
    "n_estimators": [50, 100],
    "max_depth": [5, 10, None]
}

grid = GridSearchCV(
    RandomForestClassifier(random_state=42),
    param_grid,
    cv=5,
    scoring="accuracy"
)

grid.fit(X_train_scaled, y_train)

print("Best Parameters:", grid.best_params_)
print("Best Accuracy:", grid.best_score_)
```
