## Project Objective

The objective of this project is to perform Exploratory Data Analysis (EDA) on the House Prediction dataset.

The analysis includes:

- Understanding the dataset
- Calculating summary statistics
- Visualizing data distributions
- Identifying relationships between variables
- Finding correlations among numerical features

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")
```

```python
column_names = [
    'CRIM', 'ZN', 'INDUS', 'CHAS', 'NOX', 'RM',
    'AGE', 'DIS', 'RAD', 'TAX', 'PTRATIO',
    'B', 'LSTAT', 'MEDV'
]

df = pd.read_csv(
    "house_prediction.csv",
    sep=r"\s+",
    header=None,
    names=column_names
)

df.head()
```

## Step 1: Dataset Overview

```python
print("Shape of Dataset:", df.shape)

print("\nColumn Names:")
print(df.columns)

df.info()
```

## Step 2: Summary Statistics

```python
df.describe()
```

## Step 3: Calculate Summary Statistics

```python
# Mean
print("Mean:")
print(df.mean())
```

```python
# Median
print("Median:")
print(df.median())
```

```python
# Mode
print("Mode:")
print(df.mode())
```

```python
# Standard Deviation
print("Standard Deviation:")
print(df.std())
```

### Observation

The summary statistics provide useful information about the distribution and variability of each feature in the dataset.

- Mean indicates the average value.
- Median represents the middle value.
- Mode shows the most frequent value.
- Standard deviation measures the spread of the data.

## Step 4: Histogram Visualization

```python
df.hist(figsize=(18, 12))
plt.suptitle("Histogram of House Prediction Dataset")
plt.show()
```

### Observation

Histograms show the distribution of each numerical feature and help identify skewness, spread, and potential outliers.

## Step 5: Box Plot Visualization

```python
plt.figure(figsize=(18,8))
sns.boxplot(data=df)
plt.xticks(rotation=45)
plt.title("Box Plot of Features")
plt.show()
```

### Observation

Box plots help identify the spread of data and detect potential outliers in different features.

## Step 6: Scatter Plot

```python
plt.figure(figsize=(8,6))
sns.scatterplot(x='RM', y='MEDV', data=df)

plt.title("Average Rooms vs House Price")
plt.xlabel("Average Number of Rooms (RM)")
plt.ylabel("Median House Value (MEDV)")

plt.show()
```

### Observation

The scatter plot shows a positive relationship between the average number of rooms and the median house value.

Houses with more rooms generally have higher prices.

## Step 7: Correlation Matrix

```python
correlation = df.corr()

correlation
```

## Step 8: Correlation Heatmap

```python
plt.figure(figsize=(12,10))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()
```

### Observation

The correlation heatmap helps identify relationships between variables.

- Positive values indicate a positive relationship.
- Negative values indicate an inverse relationship.
- Values close to zero indicate weak or no correlation.

# Key Insights

- The dataset contains 506 observations and 14 numerical features.
- No missing values or duplicate records were found.
- Most variables show different distributions, with some containing outliers.
- The number of rooms (RM) has a strong positive relationship with house prices (MEDV).
- LSTAT shows a negative relationship with house prices.
- Correlation analysis helps identify the most influential features affecting house values.
